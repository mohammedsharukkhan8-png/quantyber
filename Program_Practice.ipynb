```python
def caesar_encrypt(message, shift):
    s = ""
    for c in message:
        if c == ' ' or c in "0123456789":
            s += c
        elif c.isupper():
            c = ord(c)
            s += chr(c + shift if c + shift <= 90 else c - 26 + shift)
        elif c.islower():
            c = ord(c)
            s += chr(c + shift if c + shift <= 122 else c - 26 + shift)
        else:
            s += c
    return s
print(caesar_encrypt("Bloody burger 1", 3))
```

```python
def caesar_decrypt(message, shift):
    s = ""
    for c in message:
        if c == ' ' or c in "0123456789":
            s += c
        elif c.isupper():
            c = ord(c)
            s += chr(c - shift if c - shift >= 65 else c + 26 - shift)
        elif c.islower():
            c = ord(c)
            s += chr(c - shift if c - shift >= 97 else c + 26 - shift)
        else:
            s += c
    return s
#string = caesar_encrypt("Bloody burger 1", 3)
print(caesar_decrypt(caesar_encrypt("ABC", 3), 3))
```

```python
def vigenere_encrypt(message, key):
    kval = []
    s = ""
    l = len(key)
    i = l
    for c in key:
        if c.isupper():
            kval.append(ord(c) - 65)
        elif c.islower():
            kval.append(ord(c) - 97)
        else:
            kval.append(ord(c))
    for ch in message:
        if ch.isupper():
            ch = ord(ch)
            s += chr(ch + kval[l-i] if ch + kval[l-i] <= 90 else ch - 26 + kval[l-i])
            i -= 1
        elif ch.islower():
            ch = ord(ch)
            s += chr(ch + kval[l-i] if ch + kval[l-i] <= 122 else ch - 26 + kval[l-i])
            i -= 1
        else:
            s += ch
            i -= 1
            
        if i == 0:
            i = l
            
    return s
        
print(vigenere_encrypt("Hello", "CAT"))
```

```python
import os
import hashlib
def hash_password(pwd):
    salt = os.urandom(16).hex()
    combined = salt + pwd
    hashword = hashlib.sha256(combined.encode()).hexdigest()
    return salt, hashword
    
def verify_password(pwd, salt, hash):
    password = salt + pwd
    hashed = hashlib.sha256(password.encode()).hexdigest()
    
    if hashed == hash:
        return "Password verified"
    return "Wrong password Try Again!"
    
salt, hash = hash_password("Khan@07")
print(verify_password("Khan@07", salt, hash))
print(verify_password("Hack", salt, hash))
```

```python
import socket
def port_scanner(host, ports):
    open_ports = []
    for i in ports:
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        res = sock.connect_ex((host, i))
        if res == 0:
            print(i,"is open!")
            open_ports.append(i)
    if not open_ports:
        print("All primary ports are closed.")
    return

port_scanner("10.85.139.96", range(1, 65535))
```

```python
import hashlib
def hash_file(file):
    with open(file, "rb") as f:
        content = f.read()
        #print(content)
        hashed = hashlib.sha256(content).hexdigest()
    return hashed
print(hash_file("/home/irusha/Downloads/quantum_cyber_roadmap.pdf"))
```

```python
def verify_file(file,fhash):
    with open(file, "rb") as f:
        content = f.read()
        hashed = hashlib.sha256(content).hexdigest()
    if hashed == fhash:
        return "File Verified. Not tampered!"
    return "File Tampered!"
fname = "/home/irusha/Downloads/quantum_cyber_roadmap.pdf"
fhash = hash_file(fname)
print(verify_file(fname, fhash))
```
